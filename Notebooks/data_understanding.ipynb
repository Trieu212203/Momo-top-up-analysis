# Data Understanding Summary & Next Steps
After exploring the dataset, here is the summary of the data quality and the required actions for the next processing stage (**Data Cleaning**):
## 1. Transactions Table
*   **Datatype Conversion**: `Amount` is a string with commas -> Needs conversion to numeric (int).
*   **Datatype Conversion**: `Date` is stored as an object -> Needs conversion to `datetime`.
*   **Missing Values**: `Purchase_status` has 83.4% missing values -> Need a strategy to handle them (e.g., fill with "Normal" or categorize).
## 2. User Info Table
*   **Duplicates**: 38 duplicated `User_id`s found -> Needs deduplication (keeping the valid/latest record).
*   **Anomalous Dates**: `First_tran_date` contains out-of-range years (e.g., 9920) and is stored as an object -> Investigate, fix outliers, and convert to `datetime`.
*   **Inconsistent Data (Gender)**: Gender column is messy (`MALE`, `male`, `Nữ`, `Nam`, etc.) -> Standardize formatting (e.g., `Male`/`Female`).
*   **Inconsistent Data (Location)**: Overlapping categories (`Ho Chi Minh City` vs `HCMC`) -> Map to standardized names.
*   **Implicit Missing Values**: `Age` contains 2,521 `unknown` values -> Decide whether to treat as a separate cohort or impute.
## 3. Commission Table
*   **Datatype Check**: Verify if `Rate_pct` is stored as percentage percentages (e.g., `0.05` instead of `5`) to ensure accurate revenue calculation later. 
*   **Quality**: No duplicate keys found. Clean and ready to be used as a lookup table.
